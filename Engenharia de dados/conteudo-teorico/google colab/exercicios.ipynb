```python
# 1. A coluna body_mass_g está em gramas. Crie uma nova coluna chamada body_mass_kg que contenha a massa corporal convertida para quilogramas
print(df_penguins['body_mass_g'])

df_penguins['body_mass_g'] = pd.to_numeric(df_penguins['body_mass_g'], errors='coerce')
df_penguins['body_mass_kg'] = df_penguins['body_mass_g'] / 1000

print(df_penguins['body_mass_kg'])

display(df_penguins)
```

```python
#Utilizando o dataset dos pinguins limpo anteriormente, extraia e interprete as principais métricas descritivas das variáveis numéricas. O objetivo é compreender as medidas de tendência central, dispersão, amplitude e correlação.
#Exiba um resumo geral com describe().
#Calcule a média, mediana e moda da coluna body_mass_g.
#Calcule a variância, desvio padrão, mínimo, máximo e amplitude da coluna flipper_length_mm.
#Gere a matriz de correlação entre as variáveis numéricas.

df_penguins.describe()

moda = df_penguins['body_mass_g'].mode()
print(f'Moda: {moda}')

mediana = df_penguins['body_mass_g'].median()
print(f'mediana: {mediana}')

media = df_penguins['body_mass_g'].mean()
print(f'media: {media}')
```

```python
variancia = df_penguins['flipper_length_mm'].var()
print(f'variancia: {variancia}')
```

```python
desvio_padrao = df_penguins['flipper_length_mm'].std()
print(f'desvio padrao: {desvio_padrao}')
```

```python
min = df_penguins['flipper_length_mm'].min()
print(f'min: {min}')

max = df_penguins['flipper_length_mm'].max()
print(f'max: {max}')
```

```python
amplitude = max - min
print(f'amplitude: {amplitude}')
```

```python
matriz_correlacao = df_penguins.corr(numeric_only= True)
print(matriz_correlacao)
```

```python
#Utilize o dataset dos pinguins para explorar as contagens de categorias e a distribuição de valores numéricos.
#Conte quantos pinguins existem de cada espécie.
#Conte quantos pinguins existem em cada região.
#Calcule a média de massa corporal por espécie.
#Gere a distribuição de frequências da variável body_mass_g dividindo-a em intervalos.
#Analise a proporção de pinguins machos e fêmeas a partir das colunas sex_FEMALE e sex_MALE.

quantidade_penguins = df_penguins.groupby('species')['species'].count()
print(quantidade_penguins)
```

```python
quantidade_regiao = df_penguins.groupby('region')['region'].count()
print(quantidade_regiao)
```

```python
massa_corporal = df_penguins.groupby('species')['body_mass_kg'].mean()
print(massa_corporal)
```

```python
import matplotlib.pyplot as plt

#Utilize o gráfico de linhas para demonstrar as medianas do comprimento e da profundidade do bico e do comprimento da nadadeira, para cada espécie.


comprimento_bico = df_penguins.groupby('species')['culmen_length_mm'].median()
profundidade_bico = df_penguins.groupby('species')['culmen_depth_mm'].median()
comprimento_nadadeira = df_penguins.groupby('species')['flipper_length_mm'].median()

# Adicionamos o campo label='' no método plt.plot()
plt.plot(comprimento_bico.index, comprimento_bico, label='comprimento bico')
plt.plot(profundidade_bico.index, profundidade_bico, label='profundidade bico')
plt.plot(comprimento_nadadeira.index, comprimento_nadadeira, label='comprimento nadadeira')

plt.legend()  # permite a exibição das labels definidas

plt.show()
```
