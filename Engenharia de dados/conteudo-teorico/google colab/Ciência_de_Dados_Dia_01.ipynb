## 1. CONTEXTUALIZAÇÃO

### O QUE É CIÊNCIA DE DADOS?
A Ciência de Dados é a área que usa dados para gerar conhecimento e apoiar decisões. Ela combina negócios, tecnologia e estatística para entender o que aconteceu, por que aconteceu e o que pode acontecer no futuro.

O campo abrange diversas funções como Analista, Engenheiro e Cientista de Dados, que têm objetivos e responsabilidades distintos.

### Funções em Ciência de Dados

#### Analista de Dados (Foco em BI e Relatórios)
* Atua em limpeza, organização e visualização de dados.
* Foco em relatórios, dashboards e Business Intelligence (BI).
* **Habilidades:** Estatística descritiva, Storytelling, SQL, bancos relacionais e não relacionais.
* **Ferramentas:** Power BI, Google Data Studio, Python, R.

#### Engenheiro de Dados (Foco em Infraestrutura e Fluxo de Dados)
* Cria e mantém pipelines ETL/ELT e infraestruturas de dados.
* Entrega dados limpos a Analistas e Cientistas de Dados.
* Trabalha com Data Warehouses, Data Lakes e arquitetura orientada a eventos.
* **Linguagens:** Scala, Java, Python.

#### Cientista de Dados (Foco em Pesquisa e Machine Learning)
* Aplica metodologia científica, estatística inferencial e Machine Learning (ML).
* Testa hipóteses, cria modelos preditivos e gera insights.
* Trabalha com dados complexos vindos de Data Lakes.
* **Ferramentas:** Scikit-learn, TensorFlow, PyTorch.

### Ciclo de Atividades do Cientista de Dados
O processo de Ciência de Dados é um ciclo iterativo:

1.  **Ingestão de Dados**: Coleta de dados estruturados e não estruturados (manualmente, via web scraping ou dados em tempo real).
2.  **Armazenamento e Processamento**: Armazenamento em data warehouses ou data lakes. Preparação dos dados com limpeza, transformação e integração (ETL).
3.  **Análise de Dados**: Análise exploratória para entender padrões e gerar hipóteses. Uso de ferramentas e algoritmos para análises preditivas e aprendizado de máquina.
4.  **Comunicação (Apresentar Insights)**: Visualização de dados e storytelling para os stakeholders. Utilização de relatórios, gráficos e dashboards (com R e Python).

## 6. SELEÇÕES (Slicing e Filtering)

### Seleção de Colunas
* `df['nome_coluna']`: Seleciona uma única coluna (retorna uma *Series*).
* `df[['coluna1', 'coluna2']]`: Seleciona múltiplas colunas (retorna um *DataFrame*).

````python
# Exemplo: Seleção de uma única coluna
# Selecionando a coluna 'track_name'
nomes = df_spotify['track_name']
print("--- Coluna Única ('track_name') ---")
display(nomes)
````

````python
# Exemplo: Seleção de múltiplas colunas
# Selecionando as colunas 'artist_name' e 'album_name'
artista_album = df_spotify[['artist_name', 'album_name']]
print("\n--- Múltiplas Colunas ('artist_name', 'album_name') ---")
display(artista_album)
````

### Seleção de Linhas (.loc[] e .iloc[])
Existem duas formas principais de selecionar linhas:

* `.loc[]`: Seleciona pelo **rótulo (label)** do índice.
* `.iloc[]`: Seleciona pela **posição numérica** (inteiro).

Por padrão, o rótulo do índice é igual à sua posição (0, 1, 2, ...).

#### .loc[] (Seleção por Rótulo)
* `df.loc[rotulo_unico]`
* `df.loc[[rotulo1, rotulo2]]`
* `df.loc['inicio':'fim']` (Slice por rótulo **INCLUI** o valor final)

#### .iloc[] (Seleção por Posição)
* `df.iloc[posicao_unica]` (ex: `0` para a primeira linha)
* `df.iloc[[pos1, pos2]]`
* `df.iloc[inicio:fim]` (Slice por posição **EXCLUI** o valor final, como no Python)

````python
# Exemplo .loc[]
# Seleciona a linha com o rótulo de índice 3
print("--- .loc[3] ---")
display(df_spotify.loc[3])

# Seleciona as linhas com os rótulos 5, 8 e 10
print("\n--- .loc[[5, 8, 10]] ---")
display(df_spotify.loc[[5, 8, 10]])

# Seleciona as linhas com rótulos de 3 até 5 (INCLUINDO a 5)
print("\n--- .loc[3:5] (Inclusivo) ---")
display(df_spotify.loc[3:5])
````

````python
# Exemplo .iloc[]
# Seleciona a linha na posição 3 (que é a quarta linha, 0-indexada)
print("--- .iloc[3] ---")
display(df_spotify.iloc[3])

# Seleciona as linhas nas posições 5, 8 e 10
print("\n--- .iloc[[5, 8, 10]] ---")
display(df_spotify.iloc[[5, 8, 10]])

# Seleciona as linhas das posições 3 até 5 (EXCLUINDO a 5)
print("\n--- .iloc[3:5] (Exclusivo) ---")
display(df_spotify.iloc[3:5])
````

### Seleção Combinada (Linhas e Colunas)
Podemos selecionar linhas e colunas ao mesmo tempo, usando `[linhas, colunas]`:

* `.loc[]` usa **RÓTULOS** para linhas e **NOMES** para colunas.
    * `df.loc[selecao_linhas_ROTULO, selecao_colunas_NOME]`
* `.iloc[]` usa **POSIÇÕES** numéricas para linhas e colunas.
    * `df.iloc[selecao_linhas_POSICAO, selecao_colunas_POSICAO]`

````python
# Exemplos usando .iloc[]

# Coluna única (pos 3) e linha única (pos 5)
artista_pos_5 = df_spotify.iloc[5, 3] #
print(f"Valor na linha 5, coluna 3: {artista_pos_5}\n")

# Slice de Linhas (pos 2 a 4), Lista de Colunas (pos 2 e 4)
linhas_2a4_cols_pos = df_spotify.iloc[2:5, [2, 4]] #
print("--- .iloc[2:5, [2, 4]] ---")
display(linhas_2a4_cols_pos)

# Lista de Linhas (pos 0 e 10), Slice de Colunas (pos 3 e 4)
linhas_0_10_cols_pos = df_spotify.iloc[[0, 10], 3:5] #
print("\n--- .iloc[[0, 10], 3:5] ---")
display(linhas_0_10_cols_pos)

# Todas as Linhas (:), Lista de Colunas (pos 0 e 2)
linhas_todas_cols_pos = df_spotify.iloc[:, [0, 2]] #
print("\n--- .iloc[:, [0, 2]] ---")
display(linhas_todas_cols_pos.head())
````

### Localizando Valores Extremos
Para saber *onde* está o maior ou menor valor, usamos:

* `.idxmax()`: Retorna o **rótulo do índice** onde o maior valor ocorre.
* `.idxmin()`: Retorna o **rótulo do índice** onde o menor valor ocorre.

````python
# Exemplo: Encontrar a música mais longa

# 1. Encontrar o RÓTULO (índice) da música com maior 'duration_ms'
indice_mais_longa = df_spotify['duration_ms'].idxmax()
print(f"Índice (rótulo) da música mais longa: {indice_mais_longa}")

# 2. Usar .loc[] ou .iloc[] para buscar a linha inteira
# (Neste caso, .loc[indice] é mais seguro, pois idxmax retorna o rótulo)
musica_mais_longa = df_spotify.loc[indice_mais_longa]
print("\nDados da música mais longa:")
display(musica_mais_longa)
````

### Seleção Condicional (Filtros)
Consiste em filtrar linhas que atendem a critérios lógicos (filtros).

1.  Cria-se uma "máscara booleana" (uma Series de True/False).
    * Ex: `df_spotify['artist_name'] == 'Eminem'`
2.  Passa-se essa máscara para o DataFrame:
    * `df[mascara]`

Para múltiplas condições, use `&` (E) ou `|` (OU), com parênteses:
`df[(condicao1) & (condicao2)]`

````python
# Exemplo: Condição Única
# Filtrando Músicas do artista 'Clairo'
df_clairo = df_spotify[df_spotify['artist_name'] == 'Clairo']
print("--- Músicas da 'Clairo' ---")
display(df_clairo)
````

````python
# Exemplo: Múltiplas Condições (& = E)
# Filtrando Músicas do 'Radiohead' E com duração > 180000 ms

df_readiohead_longas = df_spotify[
    (df_spotify['artist_name'] == 'Radiohead') &
    (df_spotify['duration_ms'] > 180000)
]
print("\n--- Músicas do 'Radiohead' > 180000 ms ---")
display(df_readiohead_longas)
````

````python
# Exemplo: Múltiplas Condições com variáveis (mais legível)

# 1. Criar a condição para o artista
condicao_artista = df_spotify['artist_name'] == 'Clairo'

# 2. Criar a condição para a duração
condicao_duracao = df_spotify['duration_ms'] < 180000

# 3. Aplicar ambas as condições (combinadas com &)
df_clairo_curtas = df_spotify[condicao_artista & condicao_duracao]

# 4. Mostrando o Resultado
print("\n--- Músicas do 'Clairo' < 195000 ms ---")
display(df_clairo_curtas)
````

### Removendo Colunas e Linhas (.drop())
O método `.drop()` serve para remover linhas ou colunas.

Parâmetros principais:
* `labels`: Rótulo(s) a remover (nome da coluna ou rótulo do índice).
* `axis`: Define o eixo.
    * `axis=0` (padrão) = remove **LINHAS**.
    * `axis=1` = remove **COLUNAS**.
* `inplace`:
    * `False` (padrão): Retorna uma **cópia** do DataFrame com a remoção. O original não muda.
    * `True`: Modifica o DataFrame **original** e não retorna nada.

````python
# Exemplo: Remover LINHAS por rótulo

# Vamos criar uma cópia para não afetar o df original nos exemplos
df_exemplo_drop = df_spotify.copy()

# Exemplo 1a: Remover uma única linha (rótulo 3) e criar um NOVO DataFrame
print("--- DataFrame Original (head) ---")
display(df_exemplo_drop.head())

df_sem_linha_3 = df_exemplo_drop.drop(3) # axis=0 é o padrão
print("\n--- Cópia após remover linha com rótulo 3 ---")
display(df_sem_linha_3.head())

# Exemplo 1b: Remover múltiplas linhas (rótulos 0 e 2)
df_sem_linhas_0_2 = df_exemplo_drop.drop([0, 2])
print("\n--- Cópia após remover linhas 0 e 2 ---")
display(df_sem_linhas_0_2.head())

# Exemplo 1c: Remover linha (rótulo 4) DIRETAMENTE no DataFrame (inplace=True)
df_exemplo_drop.drop(4, inplace=True)
print("\n--- DataFrame original após .drop(4, inplace=True) ---")
display(df_exemplo_drop.head())
````

````python
# Exemplo: Remover COLUNAS por rótulo

# Vamos usar uma cópia limpa
df_exemplo_drop_cols = df_spotify.copy()
print("--- DataFrame Original (head) ---")
display(df_exemplo_drop_cols.head())

# Exemplo 2b: Remover múltiplas colunas e criar um NOVO DataFrame
colunas_remover = ['album_name', 'artist_track']
df_reduzido = df_exemplo_drop_cols.drop(colunas_remover, axis=1)
# Alternativa: df_reduzido = df_exemplo_drop_cols.drop(columns=colunas_remover)

print(f"\n--- Cópia após remover {colunas_remover} ---")
display(df_reduzido.head())

# Exemplo 2c: Remover uma coluna ('duration_seconds') DIRETAMENTE no original
df_exemplo_drop_cols.drop('duration_seconds', axis=1, inplace=True)
print("\n--- DataFrame original após .drop('duration_seconds', inplace=True) ---")
display(df_exemplo_drop_cols.head())
````

### Renomeando Colunas (.rename())
Usamos o `.rename()` para alterar nomes de colunas (ou índices).
Ele usa um dicionário para mapear nomes antigos para novos: `{'nome_antigo': 'nome_novo'}`.

Sintaxe: `df.rename(columns=meu_dicionario)`.

Por padrão, `.rename()` retorna uma **cópia**. Use `inplace=True` para modificar o original.

````python
# Exemplo: Renomear colunas (retornando uma cópia)

# Define o mapa de renomeação
mapa_nomes = {
    'track_name': 'musica',
    'artist_name': 'artista'
}

print("Colunas Originais:", df_spotify.columns.to_list())

# Renomeia as colunas e armazena em um NOVO DataFrame
df_spotify_renomeado = df_spotify.rename(columns=mapa_nomes)

# Mostra o resultado do NOVO DataFrame
print("Colunas do Novo DF:", df_spotify_renomeado.columns.to_list())

# Mostra que o ORIGINAL não mudou
print("Colunas Originais (não mudou):", df_spotify.columns.to_list())
````

````python
# Exemplo: Renomear colunas (inplace=True)

# Vamos usar uma cópia para este exemplo
df_spotify_copy = df_spotify.copy()
print("Colunas Originais (Cópia):", df_spotify_copy.columns.to_list())

# 1. Definir o mapa
mapa_nomes_inplace = {
    'album_name': 'album',
    'duration_ms': 'duracao_ms' # Renomeei 'duracao' para 'duracao_ms' para evitar confusão com o exercício
}

# 2. Renomear inplace (modifica df_spotify_copy diretamente)
df_spotify_copy.rename(columns=mapa_nomes_inplace, inplace=True)

# 3. Verificar as colunas (agora modificada)
print("Colunas Modificadas (Inplace):", df_spotify_copy.columns.to_list())
````
